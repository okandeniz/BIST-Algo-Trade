Robot-compatible adjusted OHLC data collection.


```python
from pathlib import Path
import sys

PROJECT_ROOT = next(path for path in (Path.cwd(), *Path.cwd().parents) if (path / "src").is_dir())
sys.path.insert(0, str(PROJECT_ROOT))

from src.config import DataConfig
from src.data_loader import (
    load_bist_tickers,
    download_robot_bundle,
    save_robot_bundle,
)
```

```python
config = DataConfig(start="2018-01-01")

tickers = load_bist_tickers(
    PROJECT_ROOT / "data" / "raw" / "bist100_sirketler.xlsx"
)

stock_prices, market_prices, errors = download_robot_bundle(tickers, config)
paths = save_robot_bundle(
    stock_prices, market_prices, errors, PROJECT_ROOT
)

print(paths)
print(stock_prices.shape, market_prices.shape, errors.shape)
```
